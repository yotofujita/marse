```python
import soundfile as sf
import matplotlib.pyplot as plt
import pyloudnorm as pyln
import numpy as np

file_list = ["Libri2Mix_test_1089-134686-0007_7021-85628-0018",
            "Libri2Mix_test_1089-134691-0019_672-122797-0023",
            "Libri2Mix_test_1089-134686-0024_1995-1836-0013",
            "Libri2Mix_test_1089-134691-0014_7127-75947-0023",
            "Libri2Mix_test_1284-1180-0020_8555-284447-0005",
            "Libri2Mix_test_1221-135766-0003_1320-122617-0009",
            "Libri2Mix_test_1284-1180-0032_2830-3979-0003",
            "Libri2Mix_test_1284-1180-0005_6829-68771-0000",
            "Libri2Mix_test_1284-1180-0014_6829-68769-0006",
            "Libri2Mix_test_121-127105-0019_4446-2275-0029",
            "Libri1Mix_test_1363-135842-0000",
            "Libri1Mix_test_1898-145720-0018",
            "Libri1Mix_test_441-130108-0041",
            "Libri1Mix_test_8238-283452-0035",
            "Libri1Mix_test_3699-175950-0005",
            "Libri1Mix_test_27-124992-0056",
            "Libri1Mix_test_229-130880-0014",
            "Libri1Mix_test_2952-410-0016",
            "Libri1Mix_test_6078-54013-0044",
            "Libri1Mix_test_201-122255-0029"]

for file in file_list:

    print(f"Processing {file}...")

    dataset = "libri2mix" if "Libri2Mix" in file else "libridemand"
    path = "./resources/audio/" + dataset + "/ConvTasNet_N1/" + file

    noisy, sr = sf.read(path + "/noisy.wav")

    clean = sf.read(path + "/clean.wav")[0]

    sf.write("/data/tmp/diff.wav", noisy - clean, samplerate=sr)

    noise = noisy - clean

    snr_power = 10 * np.log10(np.sum(clean ** 2) / np.sum(noise ** 2))
    print(f"SNR (power ratio): {snr_power:.2f} dB")

    # Same convention as LibriMix's set_loudness: ITU-R BS.1770-4 integrated
    # loudness (LUFS) per source, SNR = loudness(clean) - loudness(noise)
    meter = pyln.Meter(sr)
    loudness_clean = meter.integrated_loudness(clean)
    loudness_noise = meter.integrated_loudness(noise)

    snr = loudness_clean - loudness_noise
    print(f"SNR (LibriMix convention): {snr:.2f} dB")

    print("\n")


        
```
